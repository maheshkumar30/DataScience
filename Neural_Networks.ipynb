#**SIMPLE NEURAL NETWORK FOR REGRESSION**

```python
import numpy as np
import matplotlib.pyplot as plt
import random

```

# **ACTIVATION FUNCTIONS**


```python
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# Tanh Function
def tanh(x):
    return np.tanh(x)

# ReLU Function
def relu(x):
    return np.maximum(0, x)
```

# **SAMPLE DATA**

```python
x1 = 3
x2 = 2

# Actual Output
y_actual = 17
```

# **RANDOM WEIGHT INITIALIZATION**

```python
w1 = random.randint(1, 10)
w2 = random.randint(1, 10)

print("Initial Weight w1:", w1)
print("Initial Weight w2:", w2)
```

# **TRAINING PARAMETERS**

```python
learning_rate = 0.01
epochs = 20

# Lists to Store Values
errors = []
predictions = []
```

# **TRAINING USING FEED FORWARD AND BACK PROPAGATION**

```python
for epoch in range(epochs):

    # FEED FORWARD

    y_pred = (x1 * w1) + (x2 * w2)

    # ERROR CALCULATION

    error = (y_actual - y_pred) ** 2

    # Store Values
    errors.append(error)
    predictions.append(y_pred)

    # BACK PROPAGATION

    # Derivatives
    d_error_w1 = -2 * (y_actual - y_pred) * x1
    d_error_w2 = -2 * (y_actual - y_pred) * x2

    # Update Weights
    w1 = w1 - learning_rate * d_error_w1
    w2 = w2 - learning_rate * d_error_w2
```

**Result**

```python

    print(f"Epoch {epoch+1}")
    print(f"Predicted Output: {y_pred:.2f}")
    print(f"Error: {error:.2f}")
    print(f"Updated w1: {w1:.2f}")
    print(f"Updated w2: {w2:.2f}")
    print("-" * 40)
```

# **FINAL OUTPUT**

```python
print("\nFinal Predicted Output:", y_pred)
print("Actual Output:", y_actual)
```

# **ERROR GRAPH**

```python
plt.figure(figsize=(8,5))

plt.plot(range(1, epochs+1), errors, marker='o')

plt.title("Error Reduction Across Epochs")
plt.xlabel("Epochs")
plt.ylabel("Error")

plt.grid(True)
plt.show()
```

# **PREDICTION GRAPH**

```python
plt.figure(figsize=(8,5))

plt.plot(range(1, epochs+1), predictions, marker='o')

plt.title("Predicted Output Across Epochs")
plt.xlabel("Epochs")
plt.ylabel("Predicted y")

plt.grid(True)
plt.show()
```
